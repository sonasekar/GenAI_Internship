```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
# Activation Function: Sigmoid
# Formula: σ(x) = 1 / (1 + e^(-x))

def sigmoid(x):
    return 1 / (1 + np.exp(-x))

x = np.linspace(-10, 10, 100)
y = sigmoid(x)

plt.plot(x, y)
plt.title("Sigmoid Activation Function")
plt.grid()
plt.show()

print("Example Input:", [-2, -1, 0, 1, 2])
print("Output:", sigmoid(np.array([-2, -1, 0, 1, 2])))
```

```python
# Activation Function: Tanh
# Formula: tanh(x) = (e^x - e^(-x)) / (e^x + e^(-x))

def tanh(x):
    return np.tanh(x)

x = np.linspace(-10, 10, 100)
y = tanh(x)

plt.plot(x, y)
plt.title("Tanh Activation Function")
plt.grid()
plt.show()

print("Example Input:", [-2, -1, 0, 1, 2])
print("Output:", tanh(np.array([-2, -1, 0, 1, 2])))
```

```python
# Activation Function: ReLU
# Formula: ReLU(x) = max(0, x)

def relu(x):
    return np.maximum(0, x)

x = np.linspace(-10, 10, 100)
y = relu(x)

plt.plot(x, y)
plt.title("ReLU Activation Function")
plt.grid()
plt.show()

print("Example Input:", [-2, -1, 0, 1, 2])
print("Output:", relu(np.array([-2, -1, 0, 1, 2])))
```

```python
# Activation Function: Leaky ReLU
# Formula: f(x) = x if x > 0 else 0.01 * x

def leaky_relu(x, alpha=0.01):
    return np.where(x > 0, x, alpha * x)

x = np.linspace(-10, 10, 100)
y = leaky_relu(x)

plt.plot(x, y)
plt.title("Leaky ReLU Activation Function")
plt.grid()
plt.show()

print("Example Input:", [-2, -1, 0, 1, 2])
print("Output:", leaky_relu(np.array([-2, -1, 0, 1, 2])))
```

```python
# Activation Function: Softmax
# Formula: softmax(x_i) = e^(x_i) / Σ e^(x_j)

def softmax(x):
    exp_x = np.exp(x - np.max(x))
    return exp_x / exp_x.sum()

x = np.array([2, 1, 0.1])
print("Input:", x)
print("Softmax Output:", softmax(x))
```
