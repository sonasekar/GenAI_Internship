```python
import numpy as np
actual = np.array([10.5, 8.2, 9.8, 7.1, 11.3])
pred   = np.array([10.0, 8.7, 9.2, 7.8, 10.9])
print("Actual Values:", actual)
print("Predicted Values:", pred)
```

```python
# 1) Mean Squared Error (MSE)
mse_loss = np.mean((actual - pred)**2)
print("Mean Squared Error:", mse_loss)
```

```python
# 2) Mean Absolute Error (MAE)
mae_loss = np.mean(np.abs(actual - pred))
print("Mean Absolute Error:", mae_loss)
```

```python
# 3) Root Mean Squared Error (RMSE)
rmse_loss = np.sqrt(mse_loss)
print("RMSE:", rmse_loss)
```

```python
# 4) Binary Cross Entropy (BCE)

y_true_bin = np.array([1,0,1,0,1])
y_pred_bin = np.array([0.93,0.18,0.89,0.35,0.72])

bce_loss = -np.mean(
    y_true_bin*np.log(y_pred_bin) +
    (1-y_true_bin)*np.log(1-y_pred_bin)
)
print("Binary Cross Entropy Loss:", bce_loss)
```

```python
# 5) Categorical Cross Entropy (CCE)

y_true_cat = np.array([
[1,0,0],
[0,1,0],
[0,0,1]])

y_pred_cat = np.array([
[0.90,0.07,0.03],
[0.15,0.75,0.10],
[0.02,0.08,0.90]])

cce_loss = -np.mean(np.sum(y_true_cat * np.log(y_pred_cat), axis=1))
print("Categorical Cross Entropy Loss:", cce_loss)
```
